## Einordnung

Python hat eine andere Designphilosophie als C, C++, Perl, Lisp, Pascal, aber viele Einflüsse aus früheren Sprachen. Durch einen aktiven Community-Prozess entwickelt sich die Sprache immer weiter.

Python ist konzipiert als Sprache, die besonders gut lesbar ist, und deren Syntax es Programmierer\*innen erlaubt, Konzepte mit weniger Codezeilen auszudrücken. Klar: Python ist eine high-level-Programmiersprache, weiter weg von konkreten Maschinenmodellen als C.
Ähnlich wie Java verwendet Python eine Garbage Collection, sodass man nicht explizit Speicher reservieren und freigeben muss, um Variablen zu belegen.

Mit Python lassen sich funktionale Programmierparadigmen verwenden wie in R oder Haskell, aber auch objektorientierte Paradigmen wie in C# oder Java.
Anders als diese Sprachen ist Python *dynamisch* getypt: Der Typ hängt am Wert, nicht am Namen. So kann ein Variablenname einmal einen String zugewiesen bekommen, direkt danach einen Boolean und danach einen Integer. Wer striktere Typen bevorzugt, kann mittlerweile mit Type Annotations und speziellen Lintern (etwa `mypy`) arbeiten; der Interpreter selbst prüft sie nicht.

Es gibt zwei klassische Arten, Python zu verwenden.
Mit einer REPL (read-eval-print loop), ähnlich einer Shell in der Kommandozeile, lässt sich schnell prüfen, ob kurze Codeschnipsel syntaktisch korrekt sind und sich so verhalten, wie man erwartet.
Jede\*r Anwender\*in von Python sollte sich mit dem Interpreter (also der REPL) vertraut machen.

Durch kompilieren eines Programms (.py) zu Python Bytecode (.pyc) und ausführen des Bytecodes durch Python lassen sich komplexere Aufgaben lösen. Aus einer Datei heraus lassen sich andere als Module importieren.

### Datentypen

**Zahlen** werden entweder als Integer (int) oder als Floating Point Number (float) repräsentiert, und es gibt sogar komplexe Zahlen (complex).

Drei Dinge sind dabei anders, als man es aus C oder Java kennt. Erstens haben Python-Integer **keine feste Breite und laufen nicht über**: $2^{100}$ ist eine ganz gewöhnliche Zahl, Python vergrößert die interne Darstellung nach Bedarf. Zweitens sind `float`s Gleitkommazahlen nach **IEEE 754 binary64** (das, was C `double` nennt) — mit allem, was dazugehört, insbesondere dass die meisten Dezimalbrüche nicht exakt darstellbar sind. Drittens ist Python zwar streng, aber **dynamisch getypt**: Der Typ hängt am Wert, nicht am Variablennamen, und derselbe Name darf nacheinander verschiedene Typen tragen.

```python
print("23 ist", type(23), "und 23.0 ist", type(23.0))
print("und 23+2j ist", type(23+2j))
print("Addieren wir int und floats gibt's", type(23+23.0))
print("Division von Ganzzahlen ergibt float, also ist 5/2 =", 5/2)
print("Wenn wir das nicht wollen, können wir rechnen 5//2 =", 5//2)

print("kein Overflow:", 2**100)
print("IEEE 754:", 0.1 + 0.2 == 0.3, "denn 0.1+0.2 =", 0.1 + 0.2)
print("Modulo hat das Vorzeichen des Divisors:", 7 % 3, -7 % 3, "und dazu passt", -7 // 3)
```

Die mittlere Zeile ist der Grund, warum man Gleitkommazahlen **nie auf Gleichheit prüft**. Statt `a == b` schreibt man `abs(a - b) < 1e-9` oder, in Numpy, `np.isclose(a, b)` — das begegnet uns im ganzen Buch. Die letzte Zeile zeigt eine Eigenart, die beim Rechnen mit Indizes zählt: In Python hat `%` immer das Vorzeichen des *Divisors*, `-7 % 3` ist also $2$ und nicht $-1$ wie in C oder Java, und `//` rundet passend dazu nach unten ab statt zur Null hin.

**Strings** (Zeichenketten) sind grundsätzlich Unicode, wobei auf UTF-8 zurückgegriffen wird, wenn nichts anderes spezifiziert wird. Das bedeutet heutzutage im Alltag meist, dass die Quelltexte auch mit Umlauten und Akzentzeichen klarkommen, aber bei relativ alten Dateien (gerade aus der Python2-Ära) evtl. das Encoding geändert bzw. spezifiziert werden muss. Uns wird das nicht beschäftigen.

Anführungszeichen dürfen einfach oder doppelt sein, `'Data'` und `"Data"` sind dasselbe — praktisch, wenn im String selbst Anführungszeichen vorkommen. Drei Anführungszeichen umschließen mehrzeilige Strings. Wichtiger ist etwas anderes: Strings sind **unveränderlich** (immutable). Ein einzelnes Zeichen lässt sich nicht überschreiben, und jede Methode wie `upper()` liefert einen *neuen* String, statt den alten zu ändern. Dafür funktioniert auf Strings dasselbe Slicing wie auf Listen.


```python
s = "Data Science"
print(s[0], "|", s[5:], "|", s[-7:])   # Slicing geht auch auf Strings

try:
    s[0] = "d"
except TypeError as e:
    print("geht nicht:", e)

print(s.upper(), "| das Original ist unveraendert:", s)
```

**Wahrheitswerte und `None`.** Der Typ `bool` hat die beiden Werte `True` und `False`, und er ist ein Spezialfall von `int` — `True == 1` ist wahr. Interessanter ist, dass Python jeden Wert in einer Bedingung als Wahrheitswert lesen kann (*truthiness*): Die Zahl $0$, der leere String, die leere Liste, das leere Dictionary und `None` gelten als falsch, alles andere als wahr. Deshalb schreibt man `if liste:` statt `if len(liste) > 0:`.

`None` ist der einzige Wert vom Typ `NoneType` und spielt die Rolle, die in Java `null` hat: „hier steht nichts“. Eine Funktion ohne `return` gibt `None` zurück. Auf `None` prüft man mit `is None`, nicht mit `== None` — `is` vergleicht die Identität, und `None` gibt es nur einmal im ganzen Programm. Bei Daten begegnet uns `None` als Geschwister von `NaN` im Kapitel [Daten putzen](data-cleaning.ipynb).

```python
print(bool(0), bool(""), bool([]), bool([0]), bool(None))   # was gilt als falsch?
print("bool ist ein int:", True == 1, True + True)

x = None
if x is None:                      # nicht "x == None"
    print("x enthaelt nichts")

def ohne_return():
    pass

print("Rueckgabe ohne return:", ohne_return())
```

**Dictionaries** sind Abbildungen mit endlichem Definitionsbereich, man notiert z.B. den mathematischen Sachverhalt $f \colon \{a,b\} \to \mathbb{R},\ a \mapsto 1,\ b \mapsto 2$ und $f(a) = 1$ als

```python
f = {"a":1, "b":2}
if f == dict(a=1, b=2):
    print("Syntaktischer Zucker erlaubt es gelegentlich, auf Anführungszeichen zu verzichten.")
if f["a"] == 1:
    print("Alles wie geplant")
print(f, "is a", type(f))
try:
    print(f["c"])
    print("an diese Stelle kommt der Interpreter nie")
except KeyError:
    f["c"] = "Man muss aufpassen, wo eckige und wo runde Klammern hinkommen"
finally:
    print(f["c"]) # aber jetzt!

try:
    {[1, 2]: "geht nicht"}          # eine Liste ist veraenderlich
except TypeError as e:
    print("kein gueltiger Schluessel:", e)

print({(1, 2): "Tupel gehen"}[(1, 2)])
```

Unter der Haube ist ein Dictionary eine **Hash-Map**: Der Schlüssel wird zu einer Zahl verrechnet, und die zeigt auf den Platz, an dem der Wert liegt. Deshalb ist der Zugriff unabhängig von der Größe des Dictionaries schnell — und deshalb dürfen nur **unveränderliche** Objekte Schlüssel sein. Ein Schlüssel, der sich nachträglich ändert, läge am falschen Platz. Zahlen, Strings und Tupel gehen also, Listen und Dictionaries nicht; genau das zeigen die letzten beiden Zeilen oben.

**Tupel** sind längenfixierte unveränderliche Listen, also z.B. Paare oder Tripel. Achtung: die Addition ist nicht elementweise, sondern eine Listenkonkatenation, die ein neues (vom alten verschiedenes) Tupel erzeugt:

```python
paar = (2,3)
print(paar, "is a", type(paar), "of length", len(paar))
tripel = (4,5,6)
print(paar, "+", tripel, "=", paar+tripel)
print((0,1), "* 3 =", (0,1)*3)
```

**Mengen** sind ungeordnete Listen, auch die notiert man mit geschweiften Klammern:

```python
Menge = {3,2,1}
print(Menge)
Nochnemenge = set((2,3,1))
Menge == Nochnemenge
```

**Listen** sind geordnet und können verlängert werden. Unter der Haube ist eine Liste in CPython ein Array von Zeigern, das bei Bedarf durch ein größeres ersetzt wird. Als Anwender merkt man davon nichts und es ist für Performancefragen in der Regel irrelevant. Eine list comprehension ist eine Art, über eine Liste zu iterieren um eine weitere zu erstellen.

```python
liste = [3,1,4,1,5,9]
if 5 in liste:
    print(liste, "vom Typ", type(liste), "hat Länge", len(liste))
print(liste + [-1,-1,-1])
liste += [0]
print(liste)
tripel = (1,2,3)
print(list(tripel))
# hier kommt die list comprehension:
neue = [x+2 for x in liste]
print(neue)
```

**Slicing** nennt man es, wenn man einen Teil einer Liste in Python auswählt mit der Doppelpunkt-Syntax.

```python
zahlen = range(0,10)  # Achtung, geht mit 0 los, 10 ist nicht dabei.
print(zahlen, type(zahlen))
zahlen = list(zahlen)
print(zahlen, type(zahlen))
print(zahlen[1:9]) # hier ist auch die 9 nicht dabei.
print(zahlen[:5] + zahlen[5:])
print("es geht auch negativ:", zahlen[-1:])
print(zahlen[:-1] + zahlen[-1:])
```

**Generatoren** sind Objekte, die Listen in einer Hinsicht verallgemeinern: man kann sich das nächste Element geben lassen. Es muss aber keine ganze Liste im Speicher gehalten werden (und auch nicht im voraus berechnet werden), daher sind Generator-Ausdrücke an vielen Stellen die richtige Wahl für Datenverarbeitungs-Pipelines innerhalb Python.
Über jede Liste, jedes Tupel und jedes `range`-Objekt kann man genauso iterieren; `range` ist dabei kein Generator, sondern eine „faule“ Sequenz, die ihre Elemente ebenfalls erst bei Bedarf erzeugt.

```python
gen = (x + 2 for x in range(0,5))
print(gen, "hat den Typ", type(gen))
# und erst jetzt wird die Berechnung ausgeführt:
for x in gen:
    print(x)
```

### Gleichheit und Identität

*Wertgleichheit* wird mit dem Operator `==` geprüft,
*Referenzgleichheit* mit dem Schlüsselwort `is`,
das entspricht einer Wertgleichheit der `id`s.

```python
print(123 == 121 + 2)
a = [1,2,3]
b = a[:] # Kopie
print(id(a) == id(b))
print(a is b)
print(a is [1,2,3])
print(a == b)
```

*Typenzugehörigkeit* prüfen wir mit der Methode `isinstance`.

```python
print(isinstance("a", str))
print(isinstance(123, bool))
print(type(123))
print(isinstance(123, object))
```

Noch schwächer als `==` ist ein Vergleich der darstellenden Strings (das ist etwas anderes als eine Typenkonversion zum Typ `str`). Dabei soll `repr()` einen möglichst eindeutigen String liefern, während `str()` einen möglichst lesbaren String liefern soll. Oft lässt sich der `repr`-String als Python-Code evaluieren mit `eval`, der eine Kopie des Objekts erzeugt.

```python
a = dict(key1=1, key2=2)
b = {"key1":'1', "key2":'2'}
print("echt verschiedene dict Objekte haben verschiedene Repräsentationen:")
print("(", a, "==", b, ") =", a == b)
print("aber bei dict Objekten ist str() das gleiche wie repr()")
print("(", repr(a), "==", str(a), ") = ", repr(a) == str(a))

c = eval(repr(a))
print(c)

print(str(3) == str("3"))
print(repr(3) == repr("3"))
print(repr(3), "!=", repr("3"))
```

### Kontrollfluss, f-Strings und Dateien

Schleifen und Verzweigungen sehen in Python so aus wie unten; eine Bedingung braucht keine Klammern.

Eingerückt wird mit **vier Leerzeichen** — das ist keine Vorschrift des Interpreters, aber die des Stilhandbuchs PEP 8, und jedes Werkzeug im Ökosystem geht davon aus. Tabulatoren und Leerzeichen zu mischen ist in Python 3 ein Fehler.

Die `for`-Schleife ist immer ein *for-each*: Sie läuft über die Elemente einer Sequenz, nie über einen Zähler. Wer einen Zähler braucht, erzeugt sich mit `range(n)` eine Sequenz von Zahlen und iteriert darüber — das ist der Grund, warum `range` in Python so allgegenwärtig ist. Über ein Dictionary iteriert man mit `.items()`, dann bekommt man Schlüssel und Wert zugleich:

```python
for i in range(3):          # for-Schleife über eine Sequenz
    if i == 0:
        print(i, "ist null")
    elif i % 2 == 1:
        print(i, "ist ungerade")
    else:
        print(i, "ist gerade")

n = 5
while n > 0:                # while-Schleife
    n -= 2
print("n ist jetzt", n)

for farbe in ["blau", "rot"]:           # for-each ueber eine Liste
    print(farbe)

for k, v in {"a": 1, "b": 2}.items():   # ueber ein dict: Schluessel und Wert
    print(k, "->", v)
```

**f-Strings** setzen Werte in einen String ein. Nach einem Doppelpunkt kann man die Formatierung angeben (Breite, Ausrichtung, Nachkommastellen). Wir verwenden sie ab dem nächsten Kapitel für Ausgaben:

```python
name, anteil = "female", 0.7420382
print(f"Anteil {name}: {anteil}")
print(f"Anteil {name:>8s}: {anteil:6.2%}")   # 8 Zeichen breit, rechtsbündig; Prozent mit 2 Nachkommastellen
print(f"{2**10 = }")                         # praktisch beim Debuggen
```

**Dateien** öffnet man mit `open`, am besten in einem `with`-Block: Der sorgt dafür, dass die Datei am Ende des Blocks wieder geschlossen wird, auch wenn dazwischen ein Fehler auftritt. Über eine geöffnete Textdatei kann man zeilenweise iterieren:

```python
with open("beispiel.txt", "w", encoding="utf-8") as datei:   # "w" = schreiben
    datei.write("erste Zeile\nzweite Zeile\n")

with open("beispiel.txt", "r", encoding="utf-8") as datei:   # "r" = lesen
    for nummer, zeile in enumerate(datei, start=1):
        print(f"{nummer}: {zeile.strip()}")

import os
os.remove("beispiel.txt")   # aufräumen
```

### Objektorientierte Programmierung

### Funktionen

Eine Funktion definiert man mit `def`, der Rückgabewert kommt von `return`. Der String direkt unter der Kopfzeile ist der *Docstring*, die eingebaute Dokumentation, die `help` anzeigt. Parameter können Standardwerte haben:

```python
def quadrat(x, plus=0):
    """Quadriert x und addiert optional etwas dazu."""
    return x**2 + plus

print(quadrat(3))            # 9
print(quadrat(3, plus=1))    # 10
help(quadrat)
```

Ein Detail, das aus anderen Sprachen kommend überrascht: In Python bilden `if`, `for` und `while` **keinen** eigenen Geltungsbereich. Eine Variable, die in einem solchen Block entsteht, lebt danach weiter. Einen eigenen Geltungsbereich (*Scope*) haben nur Funktionen, Klassen und Module — was innerhalb einer Funktion zugewiesen wird, ist außerhalb nicht sichtbar.

Umgekehrt darf eine Funktion Namen von außen *lesen*, ohne sie zu übernehmen. Sobald sie einem solchen Namen etwas zuweist, gilt er für die ganze Funktion als lokal; wer wirklich die äußere Variable meint, muss sie mit `global` (oder in verschachtelten Funktionen mit `nonlocal`) anmelden. Beides braucht man selten und sollte es meiden: Eine Funktion, die von außen sichtbare Variablen verändert, ist schwer zu testen.

```python
if True:
    drinnen = 1
print("nach dem if-Block gibt es", drinnen, "- Bloecke haben keinen eigenen Scope")

def f():
    lokal = 2

f()
try:
    print(lokal)
except NameError as e:
    print("nach der Funktion dagegen nicht:", e)
```

In einer Methodendeklaration gibt es die Möglichkeit, beliebig viele Positionsargumente zuzulassen:

```python
def methode(*args):
    return args

methode()
print(methode(1,2,3,"vier"))

def neuemethode(argmusssein, *args):
    return args

print(neuemethode(1,2,3,"vier"))
```

```python
try:
    neuemethode()
except TypeError as e:
    print(e)
```

Ebenso kann man mit Schlüsselwortargumenten umgehen:

```python
def methode(**kwargs):
    return kwargs

methode()
print(methode(key="value"))

try:
    methode("123")
except TypeError as e:
    print(e)
```

Insgesamt kann man eine Folge von zwingenden Positionsargumenten, eine Folge von spezifizierten Schlüsselwortargumenten und beliebige weitere zulassen:

```python
def methode(posarg, posarg2, *args, kwarg=None, kwarg2=True, **kwargs):
    return args, kwargs

#methode() # <--- geht nicht

methode(1, 2, 3, 4, 5, kwarg2=False, kwarg3="test")
```

Um direkt ein ganzes `dict`-Objekt als Schlüsselwortargumente zu übergeben, kann man dies entpacken (und genau so Tupel):

```python
args = (1, 2, 3, 4, 5)
kwargs = {"kwarg2": False, "kwarg3": "test"}
print(methode(*args, **kwargs))

print(args)
print(*args)
```

Auch Objektattribute werden in einem `dict` verwaltet:

```python
class NeueKlasse:
    def __init__(self):
        self.x = 1
        self.y = 2
        
neues_objekt = NeueKlasse()
print(neues_objekt.__dict__)
neues_objekt.__dict__["z"] = 3
neues_objekt.z == 3
```
