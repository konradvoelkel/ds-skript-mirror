```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform

# 1. Define the grid for the 2D plot
# For both x and y, the range is [0,1] for the uniform distribution
# We'll extend slightly beyond [0,1] to clearly show where the density drops to zero.
x = np.linspace(-0.2, 1.2, 100)
y = np.linspace(-0.2, 1.2, 100)

X, Y = np.meshgrid(x, y)

# 2. Calculate the PDF values for each dimension
pdf_uniform_x = uniform.pdf(X, loc=0, scale=1)
pdf_uniform_y = uniform.pdf(Y, loc=0, scale=1)

# 3. Calculate the product density (joint PDF)
# Since X and Y are independent, their joint PDF is the product of their individual PDFs.
Z = pdf_uniform_x * pdf_uniform_y

# 4. Create the 3D surface plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface. cmap='viridis' gives a nice color gradient.
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

# 5. Set labels and title
ax.set_xlabel('X (Uniform Distribution)')
ax.set_ylabel('Y (Uniform Distribution)')
ax.set_zlabel('Density')
ax.set_title('Joint Density of Two Independent Uniform(0,1) Distributions')

# Set z-axis limits to clearly show the flat plane at Z=1 and the zero density elsewhere
ax.set_zlim(0, 1.5)

# Add a color bar to indicate density values
fig.colorbar(surf, shrink=0.5, aspect=5, label='Density')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, uniform

# 1. Define the grid for the 2D plot

# For the standard normal distribution (x-axis)
# We go from -4 to 4 standard deviations to cover most of the bell curve.
x = np.linspace(-4, 4, 100)

# For the uniform distribution (y-axis)
# We extend slightly outside [0,1] to clearly show where the density is zero.
y = np.linspace(-0.5, 1.5, 100)

# Create a 2D grid from the x and y arrays
X, Y = np.meshgrid(x, y)

# 2. Calculate the PDF values for each distribution independently

# PDF of the standard normal distribution (N(0,1))
pdf_normal = norm.pdf(X, loc=0, scale=1)

# PDF of the uniform distribution on [0,1]
# loc=0 is the lower bound, scale=1 is the range (1-0)
pdf_uniform = uniform.pdf(Y, loc=0, scale=1)

# 3. Calculate the product density (Joint PDF)
# Since the distributions are independent, their joint PDF is the product of their individual PDFs.
Z = pdf_normal * pdf_uniform

# 4. Create the 3D surface plot
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
# cmap='viridis' is a good colormap for representing data density
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

# 5. Set labels and title
ax.set_xlabel('X (Standard Normal N(0,1))')
ax.set_ylabel('Y (Uniform U([0,1]))')
ax.set_zlabel('Joint Density')
ax.set_title('Joint Density of N(0,1) x U([0,1])')

# Set z-axis limits to clearly show the range of densities
# The maximum density for N(0,1) is at x=0, which is approx 0.3989.
# The uniform component scales by 1 where it exists.
ax.set_zlim(0, 0.45)

# Add a color bar to indicate density values
fig.colorbar(surf, shrink=0.5, aspect=5, label='Density Value')

# Optional: Adjust view angle for better visualization
ax.view_init(elev=20, azim=-120) # elev (elevation angle), azim (azimuth angle)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Define the grid for the 2D plot
# For both x and y, we'll cover a range of about 4 standard deviations around the mean (0).
x = np.linspace(-4, 4, 100)
y = np.linspace(-4, 4, 100)

X, Y = np.meshgrid(x, y)

# 2. Calculate the PDF values for each dimension independently
# PDF for the standard normal distribution for X
pdf_normal_x = norm.pdf(X, loc=0, scale=2)

# PDF for the standard normal distribution for Y
pdf_normal_y = norm.pdf(Y, loc=0, scale=0.5) # mu=0, sigma=1

# 3. Calculate the product density (joint PDF)
# Since the distributions are independent, their joint PDF is the product.
Z = pdf_normal_x * pdf_normal_y

# 4. Create the 3D surface plot
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none', rstride=1, cstride=1, alpha=0.9)

# 5. Set labels and title
ax.set_xlabel('X (Normal)')
ax.set_ylabel('Y (Standard Normal)')
ax.set_zlabel('Joint Density')
ax.set_title('Joint Density of Two Independent Normal Distributions (N(0,2) x N(0,1))')

# Set appropriate limits for the z-axis for better visualization
ax.set_zlim(0, np.max(Z) * 1.1)

# Add a color bar
fig.colorbar(surf, shrink=0.5, aspect=5, label='Joint Density Value')

# Adjust the view angle for better perspective
ax.view_init(elev=50, azim=20) # elev is elevation, azim is azimuth

plt.show()
```
